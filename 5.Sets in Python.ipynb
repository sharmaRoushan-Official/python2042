### 5. Sets in Python

-----
A `set` in Python is a collection of data type used to store multiple unique value.

It is based on the mathematical concept of a set.

### Key Characteristics

1. Unordered
2. Unindexed
3. Set is Multable(can add/remove items/ unhashable)
4. Does not allow duplicate values
5. Elements must be immutable(hashable)
6. Set values called as keys

````python
s = {}
print(s,type(s))  # dictioanry
````

````python
s1 = {}  # set
s2 = {}  # by default dictionary
````

````python
s1 = {1}
print(s1)
print(type(s1))
````

````python
s1 = set()
print(s1)
print(type(s1))
````

### 2. Creating Sets

````python
s = {1,1,2,3}
print(s,type(s))
````

### 3. Duplicate Elements

sets automatically remove duplicates



````python
s = {5,3,25,6,9,9,8,7,1,1,1}
print(s)
````

````python
s1 = {5,3,25,6,9,9,8,7,1,1,1,-66,-10}
print(s1)
````

### 4. Data Types Allowed in sets

Allowed:

```text

frozenset() --> set which is immutable

````python
s1 = {10,55.63,True,False,"Hello world",(10,20,30),frozenset([10,20,30,30])}
print(s1)
````

### Not Allowed:

````python
s1 = {10,55.63,True,False,"Hello world",(10,20,30),frozenset([10,20,30,30]),{"a":'apple'}}
print(s1)
````
