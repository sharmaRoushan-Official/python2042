## 8. List methods

### 10. Sorting List
Data arrange in ascending or descending order

```python
nums = [10,20,30,40,50,50,50,60,20,60,30]
print(nums)

nums.sort() # by default ascending order
print(nums)
```

```python
nums = [10,20,30,40,50,50,50,60,20,60,30]
print(nums)

nums.reverse()
print(nums)
```

```python
nums = [10,20,30,40,50,50,50,60,20,60,30]
print(nums)

nums.sort() # by default ascending order
print(nums)

nums.reverse()
print(nums)
```

```python
nums = [10,20,30,40,50,50,50,60,20,60,30]
print(nums)

nums.sort() # by default ascending order
print(nums)

val = nums[::-1]  # reverse using slicing
print(val)
```

## 17. Built-in Function in List

1. len(): it return total number of items.
2. max(): it return maximum value in the list.
3. min(): it return minimum value in the list
