### 3. Lists in Python

----
* A `List` in python is a mutable(Can be change), Ordered, index collection that can store multiple values.
* Including mixed data types
* Lists are one of the most used data types in Python


1. List Definition

---
Syntax:

varibale = [1,2,3]

> List can be written in square bracket.

```python
my_list = [1,2,3,4]
print(my_list)
print(type(my_list))
```

```python
data = [10,20,30]
print(data,type(data))
```

### 3. List Indexing

> Positive Indexing

> Negative Indexing

```python
x = [10,25.36,"Python",True,False, [10,20],("Python",False),{10,20,30},{"a":'apple'}]
print(x)
print(type(x))
```

```python
### Positive Indexing

print(x[0])
print(x[1])
print(x[2])
print(x[3])
print(x[4])
print(x[5])
print(x[6])
print(x[7])
print(x[8])
```

```python
### Negative Indexing

x = [10,25.36,"Python",True,False, [10,20],("Python",False),{10,20,30},{"a":'apple'}]
print(x)
print(type(x))

print(x[-1])
print(x[-2])
print(x[-3])
print(x[-4])
print(x[-5])
print(x[-6])
print(x[-7])
print(x[-8])
print(x[-9])
```

### 4. Slicing of List

`var[start:end]`

```python
x = [10,25.36,"Python",True,False, [10,20],("Python",False),{10,20,30},{"a":'apple'}]
print(x)
print(type(x))
```

```python
x[2:5]
```

```python
x[0:6]  # 0-5
```

### 5. Striding of List

`var[start:end:step]`

```python
x = [10,25.36,"Python",True,False, [10,20],("Python",False),{10,20,30},{"a":'apple'}]
print(x)
print(type(x))
```

```python
x[::2]  #
```

```python
x[::3]
```
