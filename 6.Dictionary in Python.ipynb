## 6. Dictionary in Python

---
A `dictionary` is a collection of key:value paris.

* Each key is unique.
* Each key maps to a value.
* Dicationaries are mutable(unhashable/changable)
* It is written using curly braces {}

```python
student = {}
print(student,type(student))
```

```python
student = {"name":"Nitesh","age":55,"course":"Python"}
print(student)
print(type(student))
```

```python
student = {
    "name":"Nitesh",
    "age":55,
    "course":"Python"
}
print(student)
print(type(student))
```

Note:

name,age,course ---> keys of dictionary

"Nitesh",55,"Python" ---> Values of dictionary

----

### 11. Looping Through Dictionary

Loop: Iterate/travese


Loop through keys

```python
for key in student:
    print(key)
```

```python
for key in student.keys():
    print(key)
```

```python
# Valeus

for value in student.values():
    print(value)
```

```python
# items: key:value pair 

for key,value in student.items():
    print(key,":",value)
```
