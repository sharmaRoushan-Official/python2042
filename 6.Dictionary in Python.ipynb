## 6. Dictionary in Python

---
A `dictionary` is a collection of key:value paris.

* Each key is unique.
* Each key maps to a value.
* Dicationaries are mutable(unhashable/changable)
* It is written using curly braces {}

```python
student = {}
print(student,type(student))
```

```python
student = {"name":"Nitesh","age":55,"course":"Python"}
print(student)
print(type(student))
```

```python
student = {
    "name":"Nitesh",
    "age":55,
    "course":"Python"
}
print(student)
print(type(student))
```

Note:

name,age,course ---> keys of dictionary

"Nitesh",55,"Python" ---> Values of dictionary

### 3. Characteristics of Dictionary

Important points to remember
----

1. Keys must be unique



```python
data ={
    "name":"Amit", # old --- removed from memory
    "name" : "xyz" # new --> stored in memory
}
data
```

###### object data is used in backed only not in frontend - ""


Last value overwrites the previous one.

## 2. Keys must be immutable

----

Allowed:
* int,
* float,
* string,
* boolean,
* tuple,
* frozenset()


Not Allowed: Mutable

* list
* set
* dictionary

```python
data = {
    1 : "Python",  # numbers as dictionary keys
    10.25:"float value",
    "name":"String value",
    (10,20):"tuple value",
    True: "boolean value",
    False: "False value"
}
data
```
