```python
# 1. String
print("Hello world")
print("100")
print('True')
print("False")
print('10.52')

# 2. Integer
print(100)
print(1000)
print(46165)

# 3. Float
print(100.0)
print(1000.0)
print(45.26)

# 4. Boolean
print(True)
print(False)
```

type() function

```python
# 1. String
print("Hello world")
print(type("Hello world"))
print("100")
print(type("100"))
print('True')
print(type('True'))
print("False")
print('10.52')

# 2. Integer
print(100)
print(type(100))
print(1000)
print(46165)

# 3. Float
print(100.0)
print(type(100.0))
print(1000.0)
print(45.26)

# 4. Boolean
print(True)
print(type(True))
print(False)
```
