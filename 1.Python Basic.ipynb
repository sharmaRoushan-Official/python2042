## Variable and Expression in Python

1. Introduction to Variables

A `variable` is a named to store data in memory.[Primary momory-RAM]

A variable as a labelled container that hold information.



```python
name = "Bhavna"
age = 45
salary = 45000.0

print(name)
print(type(name))
print(age)
print(type(age))
print(salary)
print(type(salary))
```

Note:

* `name` stores text
* `age` stores an integer
* `salary` a decimal value
