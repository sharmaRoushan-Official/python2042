## 2. Python String

----

1. What is a string in Python?

A `string` in Python is sequence of Unicode chracters enclosed in quotes.



```python
s1 = "Hello"
s2 = 'world'
s3 = """Multi-line Comment"""

print(s1)
print(s2)
print(s3)
```

## KeyPoints:

* Strings are immutable(cannot be changed after creating)
* Indexed(Ordered Collection)   Address value 
* Supports Slicing, Iteration, and many built-in methods

## Important String Methods

---

##### Case Conversion

1. upper()
2. lower()
3. capitalize()
4. title()
5. swapcase()

```python
str1 = "hEllO wORld"
print("Original String:",str1)

### upper()

strUpr = str1.upper()
print(strUpr)

### lower()
value2 = str1.lower()
print(value2)

### capitalize()
value3 = str1.capitalize()
print(value3)

#### title()
value4 = str1.title()
print(value4)

value5 = str1.swapcase()
print(value5)
```

### Searching & Finding

----

```python
s = "Hello World"
print(s)
```

```python
print(s.find("a"))  # if character not found it will return -1
print(s.find("W")) # if chracter found it will return index of character
```

```python
print(s.index("a"))  # if chracter is not found it will return error.
```

```python
print(s.index("W"))  # if character is found it will index value.
```

```python
### checking Methods 

# isalpha()
# isdigit()
# isalnum()
# islower()
# isupper()

### It will return Boolean Value.
```

```python
s = "HEllo World"

s1 = "helloworld"

s2 = "123456"

s3 = " "

s4 = "HELLO"
print(s.isalpha())  # False extra - space
print(s1.isalpha())  # all alphabet
print(s2.isalnum())  # alphabet and number
print(s3.isspace())  # check space 
print(s4.isupper())  # upper character
print(s1.islower()) # lower check
```

### Triming: White space remove

* strip(): both side trim white space.
* lstrip(): left side trim white space.
* rstrip(): right side trim whilte space.

```python
s = "               Hello world                "
print(s)

s1 = s.strip()  # remove both side space
print(s1)

s2 = s.lstrip() # remove space from left side
print(s2)


s3 = s.rstrip()  # remove space from right side
print(s3)
```

## Replace ---> Exisiting value ---> change


```python
s = "Hello world"
print("Original Value:",s)


# var.replace("old","new")

s1 = s.replace("Hello","Hey")
print(s1)

s2 = s.replace("world","Nikhil")
print(s2)
```
