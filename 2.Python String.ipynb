## 2. Python String

----

1. What is a string in Python?

A `string` in Python is sequence of Unicode chracters enclosed in quotes.



```python
s1 = "Hello"
s2 = 'world'
s3 = """Multi-line Comment"""

print(s1)
print(s2)
print(s3)
```

## KeyPoints:

* Strings are immutable(cannot be changed after creating)
* Indexed(Ordered Collection)   Address value 
* Supports Slicing, Iteration, and many built-in methods 



## 2. Creating Strings

##### Different ways

```python
s1 = "Hello"  # double quotes
s2 = 'Hello'  # single quotes
s3 = """Hello world"""  # triple quotes

print(s1)
print(s2)
print(s3)
```

## 3. String Index:

Each character has a position or address value


1. Positive Indexing
2. Negative Indexing

```python
s = "Python"
print(s)
```

### Indexing:


|-|P|y|t|h|o|n|
|---|---|---|---|---|---|---|
|Positive|0|1|2|3|4|5|
|Negative|-6|-5|-4|-3|-2|-1|

```python
### Positive Indexing
str1 = "Python"
print(str1[0])  # var[] --> Accessing address value--> RAM
print(str1[1])
print(str1[2])
print(str1[3])
print(str1[4])
print(str1[5])
# print(str1[6])
```

```python
### Negative Indexing

str1 = "Python"

print(str1[-1])
print(str1[-2])
print(str1[-3])
print(str1[-4])
print(str1[-5])
print(str1[-6])
```

## Important String Methods

---

##### Case Conversion

1. upper()
2. lower()
3. capitalize()
4. title()
5. swapcase()

```python
str1 = "hEllO wORld"
print("Original String:",str1)

### upper()

strUpr = str1.upper()
print(strUpr)

### lower()
value2 = str1.lower()
print(value2)

### capitalize()
value3 = str1.capitalize()
print(value3)

#### title()
value4 = str1.title()
print(value4)

value5 = str1.swapcase()
print(value5)
```

### Searching & Finding

----

```python
s = "Hello World"
print(s)
```

```python
print(s.find("a"))  # if character not found it will return -1
print(s.find("W")) # if chracter found it will return index of character
```

```python
print(s.index("a"))  # if chracter is not found it will return error.
```

```python
print(s.index("W"))  # if character is found it will index value.
```
