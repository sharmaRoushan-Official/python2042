## 4. Tuples in Python

------

Q. What is a Tuple?

A `tuple` is a collection data type in Python in used to store multiple values in single variable.

Key Characteristics:

---
* Ordered Collection
* Immutable(cannot be change after creation)
* Allows duplication value
* Can store different data type
* Faster than lists
* Written using round brackets()



```python
numbers = (10,20,30,40,50)
print(numbers)
```

### Tuple: Tuple is a immutable list

```python
type(numbers)
```

```python
numbers = [10,20,30,40,50]
print(numbers)
print(type(numbers))
```

### 2. Creating Tuples

Method 1 - Using parentheses

```python
fruits = ("apple","mango")
print(fruits)
print(type(fruits))
```

Method-2 Without Parenteses

```python
fruits = "apple","mango"
print(fruits,type(fruits))
```

### Method 3: tuple constructor


```python
numbers = tuple([10,20,30,40,50])
print(numbers)
print(type(numbers))
```

### 3. Tuple with Different Data types

Tuple can store multiple data types


```python
data = ("john",10.25,True,False,[10,20,30])
print(data)
```

```python
print(type(data))
```

## 4. Nested Tuples: 
Tuple inside another tuples


```python
nested_tuple = ((1,2),(3,4),(5,6))
print(nested_tuple)
```

## 5. Accessing Tuple Elements
Tuple elements are accessed using indexing

```python
fruits = ("apple","banana","mango")
print(fruits)
```

### Positive Indexing- 0,1,2,3

### Negative Index- -1,-2,-3,-4



```python
#Positive Indexing
print(fruits[0])
print(fruits[1])
print(fruits[2])
# print(fruits[3])
```

```python
# Negative Indexing
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
```

### 6. Slicing of Elements:

Used to extract a porition of the tuple

syntax:

`tuple[start:end]`

```python
fruits = ("apple","banana","mango")
print(fruits)
```

```python
fruits[1:3]
```

```python
fruits[0:2]
```

### Striding of tuple

----

syntax:

`tuple[start:end:skip]`

```python
fruits = ("apple","banana","mango")
print(fruits)

fruits[::2]
```

```python
fruits[::3]
```

### 7. Tuple Immutability

Tuple cannot be modified after creating



```python
numbers = (10,20,30,40,50)
print(numbers)
```

```python
numbers[0] = "Nikhil"  # immutable
numbers
```

```python
numbers = [10,20,30,40,50] # mutable
print(numbers)
numbers[0] = "Nikhil"
numbers
```

### 8. Changing Tuple Elements:

Since tuples are immutable, convert into list than change the value 

```python
numbers = (10,20,30)
print(numbers)
```

```python
temp = list(numbers)
temp
```

```python
temp[2] = "Nikhil"
print(temp)
```

```python
tup1 = tuple(temp)
tup1
```

## 9. Loop Through Tuples

Using for loop


```python
fruits = ("apple","banana","mango")
print(fruits,type(fruits))
```

```python
for fruit in fruits:
    print(fruit)
```
