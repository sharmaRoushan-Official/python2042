## 4. Tuples in Python

------

Q. What is a Tuple?

A `tuple` is a collection data type in Python in used to store multiple values in single variable.

Key Characteristics:

---
* Ordered Collection
* Immutable(cannot be change after creation)
* Allows duplication value
* Can store different data type
* Faster than lists
* Written using round brackets()



````python
numbers = (10,20,30,40,50)
print(numbers)
````

### Tuple: Tuple is a immutable list

````python
type(numbers)
````

````python
numbers = [10,20,30,40,50]
print(numbers)
print(type(numbers))
````

### 2. Creating Tuples

Method 1 - Using parentheses

````python
fruits = ("apple","mango")
print(fruits)
print(type(fruits))
````

Method-2 Without Parenteses

````python
fruits = "apple","mango"
print(fruits,type(fruits))
````

## 5. Accessing Tuple Elements
Tuple elements are accessed using indexing

````python
fruits = ("apple","banana","mango")
print(fruits)
````

### Positive Indexing- 0,1,2,3

### Negative Index- -1,-2,-3,-4



````python
#Positive Indexing
print(fruits[0])
print(fruits[1])
print(fruits[2])
# print(fruits[3])
````

````python
# Negative Indexing
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
````

### 6. Slicing of Elements:

Used to extract a porition of the tuple

syntax:

`tuple[start:end]`

````python
fruits = ("apple","banana","mango")
print(fruits)
````

````python
fruits[1:3]
````

````python
fruits[0:2]
````

### Striding of tuple

----

syntax:

`tuple[start:end:skip]`

````python
fruits = ("apple","banana","mango")
print(fruits)

fruits[::2]
````

````python
fruits[::3]
````

### 7. Tuple Immutability

Tuple cannot be modified after creating



````python
numbers = (10,20,30,40,50)
print(numbers)
````

````python
numbers[0] = "Nikhil"  # immutable
numbers
````

````python
numbers = [10,20,30,40,50] # mutable
print(numbers)
numbers[0] = "Nikhil"
numbers
````

## 16. Membership Operators: `in or not in`



````python
fruits = ("apple","banana","mango")
print(fruits)
````

````python
print("Apple" in fruits)  # false
print("apple" in fruits)

print("banana" not in fruits)  # False
print("banana" in fruits)  # True
````

# 17. Tuple Methods:

Tuple have only two built-in methods

1. count() : count total number of occurrence value.
2. index(): index return index of first occurrence value.

````python
numbers = (10,5,7,8,9,3,7,3,5,6,7)
print(numbers)
````

````python
print(numbers.count(10))
````

````python
print(numbers.count(True))
````

````python
print(numbers.count(7))
print(numbers.index(7))
````

## 18. Built-in function with tuple
1.max()
2.min()
3.sorted()
4.sum()

````python
numbers = (10,5,7,8,9,3,7,3,5,6,7)
print(numbers)
````

````python
print(max(numbers))
print(min(numbers))
print(sorted(numbers))  # value arranged in ascending order.
print(sum(numbers))
````
