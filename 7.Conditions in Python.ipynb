### 4. if-elif-else ladder if


````python
# write a program to check grading system
# 90-100 ---> "A"
# 80-90 ----> "B"
# 70-80 ----> "C"
# 50-70 ----> "D"
# <50 -----> "Fail"


marks = int(input("Enter your marks: "))

if marks>=90 and marks<=100:
    print("Grade A")
elif marks>=80 and marks<90:
    print("Grade B")
elif marks>=70 and marks<80:
    print("Grade C")
elif marks>=50 and marks<70:
    print("Grade D")
elif marks<50:
    print("Fail")
else:
    print("Invalid Marks")
````

### 5. Nested if Statemenet:

condition inside another condition



````python
age = 22
citizen = False

if age>=18:
    if citizen:
        print("Elegible to vote")
    else:
        print("Not Elegible to vote: critaria not matched")
else:
    print("Underage")
````

### 10. Membership Operator - [in, not-in]

````python
fruits = ["apple","mango","Banana","carrot"]

if "Apple" in fruits:
    print("Fruit Available")
else:
    print("Fruit Not Available")
````

````python
fruits = ["apple","mango","Banana","carrot"]  # ASCII value

if "mango" in fruits:
    print("Fruit Available")
else:
    print("Fruit Not Available")
````

13. Case Statement

````python
day = int(input("Enter any number from 1-7: "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Weekday")
````
