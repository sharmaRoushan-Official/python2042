#### What is a For Loop?

A `for loop` is used to iterate over a sequence(such as list, tuple, string, dictionary, set, range) and execute a block code reapeatedly.


### Flow

1. Python takes the first item from the sequence.
2. Assigns it to the loop varaible.
3. Executes the code block
4. Move the next item
5. Repeats untill all items are processed.

Example: 1 Basic Loops

```python
fruits = ['apple',"banana","mango"]

for fruit in fruits:
    print(fruit) # apple banana mango
```

### What is a loop

A for loop is used to iterate(traverse) over a sequence and execute set of startments repeatedly

```python
for i in [1,2,3]:
    print(i)
```

### 2. Syntax of for loop

Examplanation

* for ➺ keyword
* variable(i)➺takes each value one by one.
* in ➺ keyword to access elements from sequence
* seuqence➺List, tuple, string, range, dict, set
* : ➺ terminate and start of loop body
* Indendation➺defines loop block

## 3. flow of execution

```python
for x in [10,20,30]:
    print(x)
```

Flow:

1. x = 10 ➺ print(10)
2. x = 20 ➺ print(20)
3. x = 30 ➺ print(30)

## 9. For loop with List

```python
fruits = ["apple","banana","mango"]

for fruit in fruits:
    print(fruit)
```

```python
len(fruits)
```

```python
for fruit in range(len(fruits)):
    print(fruit,fruits[fruit])
```

### 10. for loop with tuple

```python
t = (10,20,30)

for val in t:
    print(val)
```

```python
for val in range(len(t)):
    print(val,t[val])
```

### 11. for loop with set


```python
s = {1,2,2,3,4}  # unordered collection of data 

for i in s:
    print(i)
```

```python
for val in range(len(val)):
    print(val)
```

### 12. for loop with dictionary



```python
d = {'a':'apple',"b":10,"c":'cat'}

for key in d:
    print(key)
```

```python
for key in d.keys():
    print(key)
```

```python
for value in d.values():
    print(value)
```

```python
for i,j in d.items():
    print(i,":",j)
```

### 13. Using enumerate() in for loop

used when you need index + value

```python
names = ["Raj","Amit","Salooni"]

for index,name in enumerate(names):
    print(index,name)
```

```python
for i in range(len(names)):
    print(i,names[i])
```
